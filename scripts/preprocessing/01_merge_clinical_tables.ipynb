```python
import numpy as np
import pandas as pd
import os
```

Datasetinde 13 farklı csv dosyası ve toplam 1000 den fazla kolon bulunduğundan bunları %80 den fazla dolu olanları alıp tek bir csv altında topluyoruz


```python
dir_path = r"C:\Users\berkk\Desktop\Bitirme_Projesi\Csv_Files"
os.chdir(dir_path)

os.listdir()
```

```python
main_csv = None

for list in os.listdir():
    hold = os.path.join(dir_path,list)
    dataPtr = os.listdir(hold)
    dataAddress = os.path.join(hold,dataPtr[0])
    print(dataAddress)
    data = pd.read_csv(dataAddress)
    for col in data.columns:
        if data[col].isnull().mean()> 0.2:
            data = data.drop(columns=col,axis=1)
    data.drop(columns=["OASIS_session_label","days_to_visit"],inplace=True)

    data["age at visit"] = np.floor(data["age at visit"]).astype(int)


    data = data.drop_duplicates(subset=["OASISID", "age at visit"])
    if main_csv is None:
        main_csv = data
    else:
        main_csv = pd.merge(
            main_csv,
            data,
            on=["OASISID", "age at visit"],
            how="outer"
        )
    
    

main_csv.head()
```
