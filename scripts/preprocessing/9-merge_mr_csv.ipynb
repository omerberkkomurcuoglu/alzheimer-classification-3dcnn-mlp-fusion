```python
import numpy as np
import pandas as pd
import os
```

```python
data_csv = pd.read_csv("7-end1.csv")
mr_csv = pd.read_csv("Repository/4-mr_path.csv")

mr_csv["age at visit"] = np.floor(mr_csv["age at visit"]).astype(int)

mr_csv.head()

```

```python

unique = mr_csv["path"].unique()
unique
```

```python
arr_conflict = []
for i in unique:
    data_subset = mr_csv[mr_csv["path"]==i][["days_to_visit","memory","session_num"]]
    uniq_subset = data_subset["memory"].unique()
    if len(uniq_subset)>1:
        arr_conflict.append(data_subset)
```

```python
arr_conflict
```

```python
data_csv["memory"].value_counts()
```

```python
data_csv[data_csv["memory"]==3]
```

```python
mr_simply = mr_csv.drop(columns=["days_to_visit","session_num"])
mr_simply.head()
```

```python
main_csv = pd.merge(
            data_csv,
            mr_simply,
            on=["OASISID", "age at visit","memory"],
            how="outer"
        )
```

```python
main_csv.head()
```

```python
main_csv.dropna(subset=["path"],inplace=True)
main_csv.shape
```

```python
main_csv.to_csv("merge_mr1.csv")
```
