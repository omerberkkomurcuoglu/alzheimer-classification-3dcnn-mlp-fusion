```python
import os 
import shutil
```

```python
path = "C:\\Users\\berkk\\Desktop\\Bitirme_Projesi\\dataset"

os.chdir(path)
os.getcwd()
```

```python
splits = ["Train","Test"]
labels = ["0","0.5","1","2"]

for s in splits:
    splits_path = os.path.join(path,s)
    for l in labels:
        labels_path = os.path.join(splits_path,l)
        listdir_labels = os.listdir(labels_path)
        
        for dir in listdir_labels:
            flair_control = False
            t1w_control = False

            view_path = os.path.join(labels_path,dir)
            view_list = os.listdir(view_path)    
            for view in view_list:

                session_path = os.path.join(view_path,view)
                weight_path = os.path.join(session_path,"NIFTI")
                weight = os.listdir(weight_path)
                parts = weight[0].split("_")
                modality = parts[-1].replace(".nii.gz","")
                if modality == "T1w" and not t1w_control:
                    weight_path = os.path.join(weight_path,weight[0])                  
                    shutil.copy(weight_path, view_path)
                    t1w_control = True
                if modality=="FLAIR" and not flair_control:
                    weight_path = os.path.join(weight_path,weight[0])
                    shutil.copy(weight_path, view_path)

                    flair_control = True
                shutil.rmtree(session_path)


                
                
```
