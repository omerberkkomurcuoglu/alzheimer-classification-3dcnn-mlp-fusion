```python
data_splits = ["Train","Test"]
cdr = [0,0.5,1,2]
data_path = os.path.join(path,"dataset")
for i in data_splits:
    split_path = os.path.join(data_path,f"{i}")
    os.makedirs(split_path, exist_ok=True)
    print(f"Oluşturuldu: {split_path}")
    for j in cdr:
        class_path = os.path.join(split_path,f"{j}")
        os.makedirs(class_path,exist_ok=True)
        print(f"Oluşturuldu: {class_path}")


```

```python
mr_folder = os.path.join(path,"MR")
mr_folder
```

```python
listdir = os.listdir(mr_folder)
listdir[:10]
```

```python
random.shuffle(listdir)
listdir[:10]
```

```python
print(len(listdir))
value = int(len(listdir)*0.8)
value
```

```python
i = 0 
for dir in listdir:
    sample_path = os.path.join(mr_folder,dir)
    
    df_path = df[df["path"]==dir]["memory"]
    if not df_path.empty:
        memory_value = df_path.iloc[0]   # veya df_path.values[0]
        if memory_value != 0.5:
            memory_value = int(memory_value)
        
    if i<value:
        label_path = os.path.join(data_path,f"Train\\{memory_value}")
        
    else:
        label_path = os.path.join(data_path,f"Test\\{memory_value}")
        

    shutil.move(sample_path, label_path)
    print(f"{sample_path} → {label_path}")

    
    
    i+=1
        
```
